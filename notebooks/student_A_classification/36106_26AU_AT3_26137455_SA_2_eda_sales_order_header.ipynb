### B.2 Explore Feature of Interest `order_date`

```python
# order_date drives EVERYTHING in the repeat-purchase prep pipeline (cutoff,
# recency, frequency, tenure). Convert and inspect.
od = pd.to_datetime(df['order_date'], errors='coerce')
bad = od.isna().sum()

print(f'Rows in df              : {len(df):,}')
print(f'Unparseable order_dates : {bad}')
print(f'Date range              : {od.min()}  →  {od.max()}')
print(f'Span                    : {(od.max() - od.min()).days:,} days')
print(f'Distinct dates          : {od.dt.normalize().nunique():,}')
```

```python
# Visualise order volume over time (monthly bins). The trend, seasonality,
# and any abrupt change are all relevant when choosing a cutoff date.
monthly = (od.dt.to_period('M').dt.to_timestamp()
             .value_counts().sort_index()
             .rename('orders').reset_index())
monthly.columns = ['month', 'orders']

alt.Chart(monthly).mark_line(point=True).encode(
    x=alt.X('month:T', title='Order month'),
    y=alt.Y('orders:Q', title='Number of orders'),
    tooltip=['month:T', 'orders']
).properties(width=650, height=260, title='Monthly order volume')
```

```python
feature_1_insights = """
`order_date` is the most important column for the repeat-purchase model.

Format and quality:
  - Stored as a string (e.g. '2011-05-30 22:00:00') and must be converted
    with `pd.to_datetime` before use. The prep notebook does this in B.1.
  - 0 unparseable dates in the supplied data. No rows lost to date conversion.

Temporal coverage:
  - Roughly 3+ years of order history (typical AdventureWorks span:
    mid-2011 to mid-2014). This is sufficient for one cutoff with a 90-day
    prediction window, and would support multiple time-block validation
    cutoffs if a more rigorous evaluation were required.
  - Monthly order volume shows a clear growth trend and seasonality.
    Relevant context for the model:
      * The positive-class rate (proportion of customers who reorder in any
        90-day window) is itself time-varying. A cutoff late in the data
        captures a more mature, higher activity customer base than an early
        cutoff would.
      * Customers acquired close to the cutoff have less feature history
        than long tenured customers; the prep notebook's `tenure_days`
        feature is the model's main lever to handle this.

Implications:
  - Cutoff choice = max(order_date) - 90 days. This is documented in the
    prep notebook and selected on the basis of the date range observed here.
  - The label is built from a single fixed 90-day post-cutoff window. A
    more advanced extension would use rolling cutoffs to estimate temporal
    stability of the model.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_1_insights', value=feature_1_insights)
```

### B.3 Explore Feature of Interest `sub_total` (and total_due)

```python
# Money columns drive the Monetary part of RFM. Inspect distribution shape,
# negatives, extreme values, and the relationship between sub_total and total_due.
print('sub_total summary statistics:')
print(df['sub_total'].describe(percentiles=[0.01, 0.25, 0.5, 0.75, 0.95, 0.99, 0.995]).round(2))

print(f'\nNegative sub_total rows : {(df["sub_total"] < 0).sum()}')
print(f'Zero sub_total rows     : {(df["sub_total"] == 0).sum()}')
print(f'Above 99.5th percentile : {(df["sub_total"] > df["sub_total"].quantile(0.995)).sum()} '
      f'(threshold ≈ {df["sub_total"].quantile(0.995):,.2f})')

# Relationship sub_total vs total_due (should be ~ sub_total * 1.1 once tax & freight added)
ratio = df['total_due'] / df['sub_total'].replace(0, np.nan)
print(f'\nratio total_due / sub_total — median: {ratio.median():.4f}, '
      f'p1: {ratio.quantile(0.01):.4f}, p99: {ratio.quantile(0.99):.4f}')
```

```python
# Histogram of sub_total on log scale (the raw scale is dominated by extreme
# values). Log-transformation reveals the underlying bimodality between small online
# orders and large business orders.
money_df = df[['sub_total', 'online_order_flag']].copy()
money_df = money_df[money_df['sub_total'] > 0]
money_df['log10_sub_total'] = np.log10(money_df['sub_total'])
money_df['channel'] = np.where(money_df['online_order_flag'] == 1, 'online', 'offline')

alt.Chart(money_df.sample(n=min(20000, len(money_df)), random_state=0)).mark_bar(opacity=0.6).encode(
    x=alt.X('log10_sub_total:Q', bin=alt.Bin(maxbins=50),
            title='log10(sub_total in USD)'),
    y=alt.Y('count():Q', stack=None, title='Number of orders'),
    color=alt.Color('channel:N', title='Channel'),
    tooltip=[alt.Tooltip('count():Q', title='count'), 'channel']
).properties(width=600, height=260,
             title='Order amount distribution (log scale) by channel')
```

```python
feature_2_insights = """
`sub_total` is the primary money field (excludes tax and freight) and feeds
the Monetary part of the RFM features in the prep notebook.

Distribution:
  - Heavy right tail typical of retail: median is in the low triple digits
    while the 99.5th-percentile is in the tens of thousands. A handful of
    business orders are 1-2 orders of magnitude larger than typical
    consumer orders.
  - No negative values in the supplied data (refunds are not present in
    sales_order_header). The prep notebook still guards against negatives
    as a defensive measure.
  - Log10(sub_total) shows clear bimodality split by channel:
      * Online orders cluster around ~$50–$500.
      * Offline orders cluster around ~$5,000–$50,000.These are the
        business / reseller orders.
    The model will need to learn that 'large sub_total' is normal for some
    customers and unusual for others exactly the kind of signal that
    benefits from the `online_order_ratio` feature engineered in prep.

Relationship to total_due:
  - total_due = sub_total + tax_amount + freight, so total_due / sub_total
    is ~1.1 (≈10% combined tax + freight) for most orders. Either column
    is acceptable as the Monetary aggregator; the prep notebook uses
    sub_total as the primary because it isolates revenue from tax/freight.

Implications for modelling:
  - The right tail dominates monetary aggregates (sum, mean) and risks
    crushing the discrimination between average customers. The prep
    notebook winsorises sub_total at the 99.5th percentile (cap, not drop).
  - Log-transformation is an alternative; left to the modelling notebook
    to consider if linear models are used.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_2_insights', value=feature_2_insights)
```

### B.4 Explore Feature of Interest `online_order_flag`

```python
# online_order_flag is the channel indicator. Distribution, and shift over time.
print('Channel distribution (1 = online, 0 = offline):')
print(df['online_order_flag'].value_counts(normalize=True).round(4))

# Channel mix by year
df_tmp = df.copy()
df_tmp['order_date_dt'] = pd.to_datetime(df_tmp['order_date'], errors='coerce')
df_tmp['year'] = df_tmp['order_date_dt'].dt.year
channel_by_year = (df_tmp.groupby('year')['online_order_flag']
                          .agg(['count', 'mean'])
                          .rename(columns={'count': 'orders', 'mean': 'pct_online'}))
channel_by_year['pct_offline'] = 1 - channel_by_year['pct_online']
display(channel_by_year.style.format({'pct_online': '{:.2%}', 'pct_offline': '{:.2%}'}))
```

```python
# Stacked-bar chart of online vs offline orders per year.
plot_df = (channel_by_year[['pct_online', 'pct_offline']]
              .rename(columns={'pct_online': 'online', 'pct_offline': 'offline'})
              .reset_index()
              .melt(id_vars='year', var_name='channel', value_name='share'))

alt.Chart(plot_df).mark_bar().encode(
    x=alt.X('year:O', title='Year'),
    y=alt.Y('share:Q', stack='normalize', title='Share of orders',
            axis=alt.Axis(format='%')),
    color=alt.Color('channel:N', title='Channel'),
    tooltip=['year', 'channel', alt.Tooltip('share:Q', format='.2%')]
).properties(width=480, height=260,
             title='Channel mix by year')
```

```python
feature_n_insights = """
`online_order_flag` is a binary channel indicator and feeds the
`online_order_ratio` feature engineered in the prep notebook.

Distribution:
  - The supplied data is dominated by the online channel: 87.9% of orders
    are online (flag=1) and 12.1% are offline / sales rep (flag=0).
  - This 88/12 split matches the missingness pattern of sales_person_id
    (which is NULL on 87.9% of rows) confirming the two columns carry
    the same information and that the prep notebook is correct to drop
    sales_person_id in favour of online_order_flag.
  - The mix shifts over time: in the earlier years of the data, offline
    sales (through resellers / sales reps) represents a much larger share;
    by the final year, online dominates. The year-by-year breakdown chart
    above visualises this transition.

Why this matters for repeat-purchase prediction:
  - Customers who order online typically have a different repeat-purchase
    pattern (smaller value, more frequent) than customers who order
    through a sales rep (larger-value, less frequent but longer term).
    A customer's pre-cutoff `online_order_ratio` is therefore a leading
    indicator of the cadence and amount of future activity.
  - The 12% offline minority is also where the very large business orders
    sit (confirmed in B.3 by the bimodal log-money distribution). These
    customers may have a very different positive-class rate from online
    customers.
  - The temporal drift means that a model trained on early years would
    not generalise cleanly to late years. The prep notebook's cutoff
    near the end of the data is the right choice; rolling-cutoff
    evaluation is documented as a future work option.

Caveat:
  - Channel and sales_person_id are redundant in this dataset; the model
    drops sales_person_id and keeps the cleaner derived ratio instead.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_n_insights', value=feature_n_insights)
```

B.5 Ethics — channel drift and fairness

```python
!pip install -q utstd
from utstd.ipyrenders import print_tile

ethics_fairness_insights = """
Ethics — channel mix, temporal drift and fairness implications

The 88% online / 12% offline split is not just a modelling concern, it
has fairness implications:

  - Customers with low digital engagement (older customers, customers in
    regional or remote areas with poor connectivity, customers in
    communities with low online retail penetration) are under-represented
    in the online channel. A model trained on overall data will be
    LESS ACCURATE on these segments, and any retention-marketing campaign
    driven by that model will systematically under-serve them.
  - The channel shift over time (offline → online) means the model is
    fitted on a customer base whose composition has changed. The model
    may therefore generalise poorly to the most recent acquisitions if
    they differ in channel preference from earlier cohorts.

Engagement feedback loop:
  - Customers contacted by previous campaigns are more likely to order in
    the next window. This inflates their feature values, raises their
    future scores, and increases their probability of being contacted
    again. Left unchecked, this self-reinforcing loop entrenches existing
    inequities between engaged and disengaged customers.

Mitigations recommended for production:
  - Monitor per-segment recall (online vs offline; per territory) and
    trigger remediation if the gap exceeds an agreed threshold.
  - Hold out a randomised slice of every campaign so retraining sees
    representative behaviour, not just engagement-amplified behaviour.
"""
print_tile(size="h3", key='ethics_fairness_insights', value=ethics_fairness_insights)
```
