### Integration Analysis

```python
# Load order-level fact data for contextual territory analysis
try:
    sales_order_header_df = pd.read_csv(at.folder_path / "sales_order_header.csv")
except Exception as e:
    print(e)

sales_order_header_df.head()
```

```python
# Check the grain and join keys before merging
print("sales_territory rows:", len(df))
print("Unique territory_id in sales_territory:", df["territory_id"].nunique())
print("territory_id is unique in sales_territory:", df["territory_id"].is_unique)

print("\nsales_order_header rows:", len(sales_order_header_df))
print("Unique sales_order_id in sales_order_header:", sales_order_header_df["sales_order_id"].nunique())
print("sales_order_id is unique in sales_order_header:", sales_order_header_df["sales_order_id"].is_unique)

print("\nUnique territory_id in sales_order_header:", sales_order_header_df["territory_id"].nunique())
```

```python
# Join order-level sales activity to the territory dimension
territory_orders_df = sales_order_header_df.merge(
    df[["territory_id", "name", "CountryRegionCode", "group"]],
    on="territory_id",
    how="left",
    validate="many_to_one",
    indicator=True
)

# Check which order rows matched the available territory records
join_coverage = (
    territory_orders_df["_merge"]
    .value_counts(dropna=False)
    .rename_axis("join_status")
    .reset_index(name="order_rows")
)

join_coverage["percentage"] = (
    join_coverage["order_rows"] / len(territory_orders_df) * 100
).round(2)

join_coverage
```

```python
# Keep orders that successfully matched to the available territory dimension rows
matched_territory_orders_df = territory_orders_df[
    territory_orders_df["_merge"] == "both"
].copy()

matched_territory_orders_df.shape
```

```python
# Convert dates for summary analysis
matched_territory_orders_df["order_date"] = pd.to_datetime(
    matched_territory_orders_df["order_date"],
    errors="coerce"
)

# Summarise order activity for the available territories
territory_order_summary = (
    matched_territory_orders_df
    .groupby(["name", "CountryRegionCode", "group"], as_index=False)
    .agg(
        order_count=("sales_order_id", "nunique"),
        customer_count=("customer_id", "nunique"),
        total_sales_value=("total_due", "sum"),
        average_order_value=("total_due", "mean"),
        online_order_share=("online_order_flag", "mean"),
        first_order_date=("order_date", "min"),
        last_order_date=("order_date", "max")
    )
)

territory_order_summary["online_order_share"] = (
    territory_order_summary["online_order_share"] * 100
).round(2)

territory_order_summary.sort_values("total_sales_value", ascending=False)
```

```python
alt.Chart(territory_order_summary).mark_bar().encode(
    x=alt.X("name:N", title="Sales Territory", sort="-y"),
    y=alt.Y("total_sales_value:Q", title="Total Order Value"),
    color=alt.Color("group:N", title="Territory Group"),
    tooltip=[
        "name",
        "CountryRegionCode",
        "group",
        alt.Tooltip("total_sales_value:Q", format=",.2f"),
        alt.Tooltip("average_order_value:Q", format=",.2f"),
        "order_count"
    ]
).properties(
    title="Matched Total Order Value by Territory",
    width=500,
    height=300
)
```

```python
territory_integration_insights = """
The sales_territory table was integrated with sales_order_header to understand whether the territory dimension can provide useful context for transaction-level analysis.

The grain of the two tables is different. sales_territory contains one row per territory, with 4 rows and 4 unique territory_id values. sales_order_header contains one row per sales order, with 31,465 rows and 31,465 unique sales_order_id values. Therefore, the appropriate join is a many-orders-to-one-territory join using territory_id.

The join coverage is incomplete. Only 15,357 sales order header rows, or 48.81%, match one of the four available territory records. The remaining 16,108 rows, or 51.19%, refer to territory_id values that are not present in the sales_territory extract. This means the available sales_territory table does not cover all order territories in the transactional data.

For the matched territories, Australia has the highest matched total order value and order volume. The United Kingdom and France also contribute substantial matched sales value, while Germany has the lowest matched total order value among the four available territories. France has a higher average order value than Australia and the United Kingdom, despite having fewer matched orders.

This integration analysis shows that sales_territory can provide useful regional context, but it should not be used as a primary clustering base. The table contains only four territory records and has incomplete coverage against sales_order_header. For the selected product-clustering use case, territory information may be considered only as optional contextual information if product-level regional features are later engineered from complete and reliable transaction joins.
"""
```

```python
print_tile(size="h3", key='territory_integration_insights', value=territory_integration_insights)
```
