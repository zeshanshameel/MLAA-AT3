### B.4 Explore Feature of Interest `sales_growth_pct`

```python
# Create and explore derived feature: sales_growth_pct
df["sales_growth_pct"] = (
    (df["sales_ytd"] - df["sales_last_year"]) / df["sales_last_year"]
) * 100

df[["name", "CountryRegionCode", "group", "sales_ytd", "sales_last_year", "sales_growth_pct"]].sort_values(
    by="sales_growth_pct",
    ascending=False
)
```

```python
alt.Chart(df).mark_bar().encode(
    x=alt.X("name:N", title="Sales Territory", sort="-y"),
    y=alt.Y("sales_growth_pct:Q", title="Sales Growth (%)"),
    color=alt.Color("group:N", title="Territory Group"),
    tooltip=[
        "name",
        "CountryRegionCode",
        "group",
        alt.Tooltip("sales_ytd:Q", format=",.2f"),
        alt.Tooltip("sales_last_year:Q", format=",.2f"),
        alt.Tooltip("sales_growth_pct:Q", format=".2f")
    ]
).properties(
    title="Sales Growth Percentage by Territory",
    width=500,
    height=300
)
```

```python
feature_3_insights = """
The third feature explored is sales_growth_pct, which was created from sales_ytd and sales_last_year.
It measures the percentage change between current year-to-date sales and last year sales for each territory.
This derived feature is useful because it gives more insight than looking at current or previous sales separately.

The United Kingdom has the highest sales growth percentage at approximately 206.45%, followed by Germany at approximately 190.93%.
Australia also shows strong growth at approximately 162.35%, while France has the lowest growth at approximately 99.14%.
Although France had the highest sales last year, its growth rate is the lowest among the four territories, while the United Kingdom shows the strongest improvement.

This feature has no missing values because both sales_ytd and sales_last_year are complete and non-zero.
However, the result should be interpreted carefully because the dataset contains only four territories.
For clustering, sales_growth_pct could be useful as a regional context feature, but it should be combined with other customer, product, sales, cost, and promotion features to create meaningful segments.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_3_insights', value=feature_3_insights)
```
