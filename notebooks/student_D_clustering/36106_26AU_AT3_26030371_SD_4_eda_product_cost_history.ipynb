### B.3 Explore Feature of Interest `end_date`

```python
# Create parsed date copy for cost-period analysis
cost_history_df = df.copy()

cost_history_df["start_date"] = pd.to_datetime(
    cost_history_df["start_date"],
    errors="coerce"
)

cost_history_df["end_date"] = pd.to_datetime(
    cost_history_df["end_date"],
    errors="coerce"
)

cost_history_df["is_current_cost_record"] = cost_history_df["end_date"].isna()

cost_history_df["is_current_cost_record"].value_counts()
```

```python
cost_period_summary = (
    cost_history_df
    .groupby(["start_date", "is_current_cost_record"], as_index=False)
    .agg(
        cost_record_count=("product_id", "count"),
        product_count=("product_id", "nunique")
    )
)

cost_period_summary
```

```python
alt.Chart(cost_period_summary).mark_bar().encode(
    x=alt.X("start_date:T", title="Cost Period Start Date"),
    y=alt.Y("cost_record_count:Q", title="Number of Cost Records"),
    color=alt.Color(
        "is_current_cost_record:N",
        title="Current Cost Record"
    ),
    tooltip=[
        "start_date",
        "is_current_cost_record",
        "cost_record_count",
        "product_count"
    ]
).properties(
    title="Cost History Records by Start Period",
    width=550,
    height=300
)
```

```python
feature_2_insights = """
The second feature explored is end_date because its missingness has business meaning in a product cost history table. In this dataset, 200 cost-history records have a populated end_date, while 195 records have a missing end_date.

The start-period summary shows a clear structure. The 72 records starting on 2011-05-31 and the 128 records starting on 2012-05-30 have closed cost periods with end dates. In contrast, all 195 records starting on 2013-05-30 have no end date and are marked as current cost records.

This pattern indicates that missing end_date values are not ordinary missing-data errors. They represent open or active cost records for products whose latest observed cost period has not been closed in the extract. Therefore, these rows should be retained and interpreted carefully during data preparation.

For the selected product segmentation use case, this distinction matters because end_date can help identify current/open cost records when deriving product-level cost features. However, the raw date fields should not be used directly as clustering features without a clear business transformation."""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_2_insights', value=feature_2_insights)
```

### B.4 Explore Feature of Interest `cost_history_count`

```python
# Aggregate historical cost records to product level
product_cost_summary = (
    cost_history_df
    .groupby("product_id", as_index=False)
    .agg(
        cost_history_count=("standard_cost", "count"),
        min_standard_cost=("standard_cost", "min"),
        max_standard_cost=("standard_cost", "max"),
        avg_standard_cost=("standard_cost", "mean"),
        first_cost_start_date=("start_date", "min"),
        latest_cost_start_date=("start_date", "max")
    )
)

product_cost_summary["standard_cost_range"] = (
    product_cost_summary["max_standard_cost"] -
    product_cost_summary["min_standard_cost"]
)

product_cost_summary.head()
```

```python
cost_history_count_distribution = (
    product_cost_summary["cost_history_count"]
    .value_counts()
    .sort_index()
    .reset_index()
)

cost_history_count_distribution.columns = [
    "cost_history_count",
    "product_count"
]

cost_history_count_distribution
```

```python
alt.Chart(cost_history_count_distribution).mark_bar().encode(
    x=alt.X("cost_history_count:O", title="Cost Records per Product"),
    y=alt.Y("product_count:Q", title="Number of Products"),
    tooltip=["cost_history_count", "product_count"]
).properties(
    title="Distribution of Cost History Records per Product",
    width=450,
    height=300
)
```

```python
product_cost_summary["standard_cost_range"].describe()
```

```python
feature_3_insights = """
The third feature explored is cost_history_count, which measures how many historical cost records are available for each product after summarising the cost history table to product level.

The product-level summary contains 293 unique products. Of these, 216 products have only 1 cost-history record, 52 products have 2 records, and 25 products have 3 records. This confirms that product_cost_history is not already at one row per product and that some products appear repeatedly because their standard cost is tracked across historical periods.

The standard_cost_range feature gives an additional view of observed cost variation over time. The median cost range is 0, meaning at least half of the products show no cost change within the available history records. However, the maximum range is approximately 234.26, showing that a smaller group of products experienced larger cost movement.

This relationship is important for product clustering because raw cost-history rows cannot be used directly in a one-row-per-product modelling table. Instead, the history should be represented through product-level features such as latest available standard cost, average standard cost, cost_history_count, and standard_cost_range, depending on the business purpose and available cost coverage.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_3_insights', value=feature_3_insights)
```
