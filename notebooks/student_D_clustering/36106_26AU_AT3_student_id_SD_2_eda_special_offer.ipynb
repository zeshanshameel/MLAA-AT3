### B.2 Explore Feature of Interest `discount_pct`

```python
# Explore feature of interest: discount_pct
feature_name = "discount_pct"

df[[feature_name]].describe()
```

```python
# Sort offers by discount percentage
df[["description", "type", "category", "min_quantity", "max_quantity", "discount_pct"]].sort_values(
    by="discount_pct",
    ascending=False
)
```

```python
alt.Chart(df).mark_bar().encode(
    x=alt.X("description:N", title="Special Offer", sort="-y"),
    y=alt.Y("discount_pct:Q", title="Discount Percentage"),
    color=alt.Color("category:N", title="Offer Category"),
    tooltip=[
        "description",
        "type",
        "category",
        alt.Tooltip("min_quantity:Q", title="Minimum Quantity"),
        alt.Tooltip("max_quantity:Q", title="Maximum Quantity"),
        alt.Tooltip("discount_pct:Q", format=".2f", title="Discount Percentage")
    ]
).properties(
    title="Discount Percentage by Special Offer",
    width=650,
    height=350
)
```

```python
feature_1_insights = """
The first feature explored is discount_pct, which represents the discount percentage applied by each special offer.
This feature is important because discount depth can influence customer purchasing behaviour, product demand, and promotion sensitivity.

The discount_pct values range from 0.00 to 0.50, meaning the offers range from no discount to a maximum discount of 50%.
The average discount is 0.22, or 22%, while the median is 0.175, or 17.5%.
The highest discounts are linked to offers such as Half-Price Pedal Sale and Mountain Tire Sale, both at 50%.
There is also one No Discount offer with a discount value of 0.00.

The chart shows that customer-focused offers have some of the highest discount percentages, while reseller offers cover a wider range of discounts from low volume discounts to clearance and overstock offers.
There are no missing values in discount_pct, so no imputation is required for this feature.
For clustering, discount_pct can help create promotion-related features such as maximum discount received, average discount exposure, and discount sensitivity."""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_1_insights', value=feature_1_insights)
```

### B.3 Explore Feature of Interest `type`

```python
# Explore feature of interest: type
feature_name = "type"

df[feature_name].value_counts()
```

```python
# Compare offer types with discount statistics
df.groupby("type").agg(
    offer_count=("special_offer_id", "count"),
    avg_discount_pct=("discount_pct", "mean"),
    min_discount_pct=("discount_pct", "min"),
    max_discount_pct=("discount_pct", "max"),
    avg_min_quantity=("min_quantity", "mean")
).sort_values(by="offer_count", ascending=False)
```

```python
alt.Chart(df).mark_bar().encode(
    x=alt.X("type:N", title="Offer Type", sort="-y"),
    y=alt.Y("count():Q", title="Number of Offers"),
    color=alt.Color("type:N", legend=None),
    tooltip=["type", "count()"]
).properties(
    title="Number of Special Offers by Type",
    width=500,
    height=300
)
```

```python
feature_2_insights = """
The second feature explored is type, which describes the type of promotion used for each special offer.
This feature is important because it shows the retailer's promotion strategy and helps explain why different discounts may exist.

The most common offer type is Volume Discount, with 5 offers. Seasonal Discount and Excess Inventory each appear 3 times, while Discontinued Product and New Product each appear 2 times.
There is also 1 No Discount offer. This shows that the retailer uses a mix of promotional strategies, including quantity-based discounts, seasonal sales, inventory clearance, discontinued product discounts, and new product promotions.

The grouped statistics show that Excess Inventory has the highest average discount at about 38.33%, followed by Discontinued Product at 37.5%.
Volume Discount offers are more frequent but have a lower average discount of about 10.4%, and they also have higher minimum quantity requirements.
This suggests that different offer types serve different business purposes: some are used to clear stock, while others encourage larger purchases.

There are no missing values in the type column. For clustering, this feature can support promotion-related variables such as offer frequency by type or whether customers/products are mainly exposed to clearance, seasonal, or volume-based offers."""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_2_insights', value=feature_2_insights)
```

### B.4 Explore Feature of Interest `category`

```python
# Explore feature of interest: category
feature_name = "category"

df[feature_name].value_counts()
```

```python
# Compare offer categories with discount statistics
df.groupby("category").agg(
    offer_count=("special_offer_id", "count"),
    avg_discount_pct=("discount_pct", "mean"),
    min_discount_pct=("discount_pct", "min"),
    max_discount_pct=("discount_pct", "max"),
    avg_min_quantity=("min_quantity", "mean")
).sort_values(by="offer_count", ascending=False)
```

```python
alt.Chart(df).mark_bar().encode(
    x=alt.X("category:N", title="Offer Category", sort="-y"),
    y=alt.Y("count():Q", title="Number of Offers"),
    color=alt.Color("category:N", legend=None),
    tooltip=["category", "count()"]
).properties(
    title="Number of Special Offers by Category",
    width=450,
    height=300
)
```

```python
feature_3_insights = """
The third feature explored is category, which identifies the target category of each special offer.
This feature is useful because it shows whether promotions are mainly aimed at resellers, customers, or no-discount situations.

The majority of offers are in the Reseller category, with 13 out of 16 offers.
Customer offers appear only 2 times, and there is 1 No Discount category.
This shows that the retailer's special offer strategy is mainly focused on reseller promotions rather than direct customer promotions.

The grouped statistics show that Customer offers have the highest average discount percentage at 50%, although there are only two customer offers.
Reseller offers have a lower average discount of about 19.38%, but they are much more frequent and include a wider range of discount levels from 2% to 40%.
The Reseller category also has a higher average minimum quantity, which suggests that reseller promotions are often designed to encourage larger purchase volumes.

There are no missing values in the category column.
For clustering, this feature can help create promotion exposure features, such as whether products or customers are mainly associated with reseller discounts, customer discounts, or no-discount purchases."""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_3_insights', value=feature_3_insights)
```
