### Integration Analysis: Special Offer Usage in Sales Order Lines

```python
# Load order-line fact data for contextual special offer analysis
try:
    sales_order_detail_df = pd.read_csv(at.folder_path / "sales_order_detail.csv")
except Exception as e:
    print(e)

sales_order_detail_df.head()
```

```python
# Check table levels and join keys before merging
print("special_offer rows:", len(df))
print("Unique special_offer_id in special_offer:", df["special_offer_id"].nunique())
print("special_offer_id is unique in special_offer:", df["special_offer_id"].is_unique)

print("\nsales_order_detail rows:", len(sales_order_detail_df))
print("Unique sales_order_detail_id in sales_order_detail:", sales_order_detail_df["sales_order_detail_id"].nunique())
print("sales_order_detail_id is unique in sales_order_detail:", sales_order_detail_df["sales_order_detail_id"].is_unique)

print("\nUnique special_offer_id used in sales_order_detail:", sales_order_detail_df["special_offer_id"].nunique())
```

```python
# Join order-line sales activity to special offer details
offer_sales_df = sales_order_detail_df.merge(
    df[
        [
            "special_offer_id",
            "description",
            "type",
            "category",
            "discount_pct",
            "min_quantity",
            "max_quantity"
        ]
    ],
    on="special_offer_id",
    how="left",
    validate="many_to_one",
    indicator=True
)

join_coverage = (
    offer_sales_df["_merge"]
    .value_counts(dropna=False)
    .rename_axis("join_status")
    .reset_index(name="order_line_rows")
)

join_coverage["percentage"] = (
    join_coverage["order_line_rows"] / len(offer_sales_df) * 100
).round(2)

join_coverage
```

```python
# Summarise how offers appear in order-line transactions
offer_usage_summary = (
    offer_sales_df
    .groupby(
        ["description", "type", "category", "discount_pct"],
        as_index=False
    )
    .agg(
        order_line_count=("sales_order_detail_id", "nunique"),
        order_count=("sales_order_id", "nunique"),
        product_count=("product_id", "nunique"),
        total_quantity=("order_quantity", "sum"),
        total_line_value=("line_total", "sum")
    )
)


offer_usage_summary.sort_values("order_line_count", ascending=False)
```

```python
alt.Chart(offer_usage_summary).mark_bar().encode(
    x=alt.X("description:N", title="Special Offer", sort="-y"),
    y=alt.Y("order_line_count:Q", title="Number of Order Lines"),
    color=alt.Color("type:N", title="Offer Type"),
    tooltip=[
        "description",
        "type",
        "category",
        alt.Tooltip("discount_pct:Q", format=".2f", title="Offer Discount"),
        "order_line_count",
        "order_count",
        "product_count",
        alt.Tooltip("total_line_value:Q", format=",.2f")
    ]
).properties(
    title="Special Offer Usage in Sales Order Lines",
    width=700,
    height=350
)
```

```python
# Visualise promotional offer usage excluding the dominant No Discount rule
promotional_offer_usage = offer_usage_summary[
    offer_usage_summary["description"] != "No Discount"
].copy()

alt.Chart(promotional_offer_usage).mark_bar().encode(
    x=alt.X("description:N", title="Promotional Offer", sort="-y"),
    y=alt.Y("order_line_count:Q", title="Number of Order Lines"),
    color=alt.Color("type:N", title="Offer Type"),
    tooltip=[
        "description",
        "type",
        "category",
        alt.Tooltip("discount_pct:Q", format=".2f", title="Offer Rule Discount"),
        "order_line_count",
        "order_count",
        "product_count",
        alt.Tooltip("total_line_value:Q", format=",.2f")
    ]
).properties(
    title="Promotional Offer Usage in Sales Order Lines",
    width=700,
    height=350
)
```

```python
special_offer_usage_insights = """
Because special_offer is a small offer-level dimension table, it was linked to sales_order_detail to examine how offer rules appear in actual order-line transactions.

The grain changes across the join. special_offer has one row per offer rule, supported by 16 rows and 16 unique special_offer_id values. sales_order_detail has one row per sales order line, supported by 42,100 rows and 42,100 unique sales_order_detail_id values. Therefore, this is a many-order-lines-to-one-offer join using special_offer_id.

The join coverage is complete: all 42,100 order lines matched to a record in the special_offer table. However, only 10 of the 16 available special offers appear in sales_order_detail, showing that not every offer rule in the dimension is used in the observed transactional extract.

The offer usage distribution is highly imbalanced. The No Discount rule is linked to 40,081 order lines, which is about 95.2% of all sales order detail rows. Among promotional offers, Volume Discount 11 to 14 is the most frequently linked offer, followed by Touring-3000 Promotion and Volume Discount 15 to 24.

Because the No Discount rule dominates the full usage chart, a second view excluding No Discount was used to compare the relative usage of promotional offer rules more clearly. This second view shows that promotional activity is not evenly distributed across offer types: volume discounts and selected new-product promotions appear much more frequently in the observed order lines than several clearance, seasonal, and excess-inventory offers.

The joined analysis shows how special offer rules appear in observed order-line transactions. The full distribution is dominated by the No Discount rule, while promotional offers are much less frequent. This means promotion-related modelling features may still be useful, but their imbalance should be considered during later product-level feature engineering and cluster interpretation.

This analysis supports the selected product segmentation use case because offer type, discount depth, and promotion exposure may help distinguish products that sell mainly without promotional offers from products that are more promotion-linked.
"""
```
