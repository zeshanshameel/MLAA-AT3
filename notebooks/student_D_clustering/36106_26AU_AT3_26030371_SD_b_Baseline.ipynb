```python
baseline_performance_explanations = """
No supervised baseline model performance has been computed for this task. Because this is an
unsupervised clustering problem, there is no target variable and no ground-truth label against
which predictions can be compared. Standard supervised performance metrics such as accuracy,
precision, recall, F1 score, MAE, RMSE, and R-squared all require a known outcome to measure
against, which is not available here. A baseline performance figure in the supervised sense is
therefore undefined, and attempting to report one would be methodologically incorrect.

Instead, clustering performance in the modelling notebook (SD-c) will be evaluated using internal
validation evidence and business interpretability. Inertia will be tracked across different cluster
counts and the elbow method will be used to identify where additional clusters stop providing
meaningful reductions in within-cluster distance. Silhouette score will measure how well each
product fits within its assigned cluster compared with the nearest alternative cluster, where
higher values indicate stronger separation and cohesion. Davies-Bouldin index will be used as a
supplementary measure of cluster compactness and inter-cluster separation, with lower values
preferred. Cluster size balance will also be reviewed because heavily imbalanced clusters reduce
business usefulness.

For a clustering solution to be considered meaningful in this context, it must satisfy both
technical and business requirements. Technically, the selected cluster count should be supported
by inertia trends and internal validation metrics rather than chosen arbitrarily. From a business
perspective, each cluster must be interpretable in terms of margin-aware promotion decisions. This
means the product segments should differ meaningfully across demand volume, gross margin,
promotional eligibility, and actual promotional usage so that distinct promotional strategies can
be recommended for each group.

A solution that produces acceptable internal scores but creates clusters that are indistinguishable
in business terms would not satisfy the objective of this assessment. The final model will therefore
be judged by the combination of clustering evidence, cluster balance, and actionable business
interpretation.
"""
```
